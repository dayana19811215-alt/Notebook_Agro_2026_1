### Ejemplos sencillos de Python para Agroindustria

Aquí tienes tres scripts básicos que ilustran tareas comunes:

#### 1. Cálculo del Índice de Cosecha

Este código ayuda a determinar qué tan eficiente fue la producción de un cultivo comparando el peso del producto comercial frente a la biomasa total.

```
# Datos de entrada
peso_fruto = 500  # kg
biomasa_total = 1200  # kg

# Cálculo
indice_cosecha = (peso_fruto / biomasa_total) * 100

print(f"El Índice de Cosecha es del {indice_cosecha:.2f}%")
```

#### 2. Monitoreo de Temperatura en Almacén

Imagina un sensor que envía datos. Este código evalúa si la temperatura es segura para conservar, por ejemplo, frutas frescas.

```
temperatura_actual = 12  # Grados Celsius

if temperatura_actual > 10:
    print("¡Alerta! La temperatura excede el límite de conservación.")
elif temperatura_actual < 2:
    print("¡Alerta! Riesgo de daño por frío.")
else:
    print("Temperatura óptima. El producto está seguro.")
```

#### 3. Conversión de Unidades (Grados Brix)

Los grados Brix indican el porcentaje de sólidos disueltos (azúcar) en una solución. Este script estima el contenido de azúcar en gramos para un volumen dado de jugo.

```
volumen_jugo = 5.0  # Litros
grados_brix = 12.0  # % de azúcar p/p (aprox)

# Cálculo rápido: (Brix / 100) * Volumen * Densidad (asumiendo dens. cercana a 1)
azucar_total = (grados_brix / 100) * volumen_jugo * 1000

print(f"En {volumen_jugo}L de jugo hay aproximadamente {azucar_total}g de azúcar.")
```
